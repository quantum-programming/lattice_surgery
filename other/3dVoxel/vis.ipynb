```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LightSource
from mpl_toolkits.mplot3d import Axes3D


def parse_output(input_str):
    input_lines = input_str.strip().splitlines()
    # チップの寸法
    w, h, _l = map(int, input_lines[0].split())
    input_lines.pop(0)

    # 量子ビット数
    n1, n2 = map(int, input_lines[0].split())
    total = n1 + n2
    input_lines.pop(0)

    # 量子ビットの位置
    xyz = []
    for _ in range(total):
        x, y, z = map(int, input_lines[0].split())
        input_lines.pop(0)
        xyz.append((x, y, z))

    m, max_turn = map(int, input_lines[0].split())
    max_turn += 1
    input_lines.pop(0)

    grid = np.zeros((max_turn, h, w), dtype=int)
    for i, (x, y, z) in enumerate(xyz):
        if i < n1:
            grid[z, x, y] = -1
        else:
            grid[z, x, y] = -2

    for task_id in range(m):
        _cnt = int(input_lines[0])
        input_lines.pop(0)
        _t_inst = list(map(int, input_lines[0].split()))
        input_lines.pop(0)

        path_len = int(input_lines[0])
        input_lines.pop(0)
        for _ in range(path_len):
            t, x, y, z = map(int, input_lines[0].split())
            input_lines.pop(0)
            assert 1 <= t <= max_turn
            assert 1 <= x <= w
            assert 1 <= y <= h
            assert z == 0
            grid[t - 1, x - 1, y - 1] = task_id + 1  # タスクIDは1から始まる

    return grid


def vis(grid: np.ndarray, figname: str = "vis.png"):
    color_voxels = np.empty((grid.shape[1], grid.shape[2], grid.shape[0]), dtype=object)

    for i in range(grid.shape[0]):
        for j in range(grid.shape[1]):
            for k in range(grid.shape[2]):
                if grid[i, j, k] > 0:
                    color_voxels[j, k, i] = f"C{grid[i, j, k] % 10 + 1}"

    ax: Axes3D = plt.figure().add_subplot(projection="3d")  # type: ignore
    ax.get_xaxis().set_ticks([])
    ax.get_yaxis().set_ticks([])
    ax.get_zaxis().set_ticks([])
    ax.xaxis.line.set_color((0, 0, 0, 0))  # type: ignore
    ax.yaxis.line.set_color((0, 0, 0, 0))  # type: ignore
    ax.zaxis.line.set_color((0, 0, 0, 0))  # type: ignore
    ax.xaxis.pane.set_edgecolor("black")  # type: ignore
    ax.yaxis.pane.set_edgecolor("black")  # type: ignore
    ax.zaxis.pane.set_edgecolor("black")  # type: ignore
    ax.grid(False)
    # ax.set_xlim3d(xmin=0, xmax=grid.shape[2] - 1)
    # ax.set_ylim3d(ymin=0, ymax=grid.shape[1] - 1)
    # ax.set_zlim3d(zmin=0, zmax=grid.shape[0] - 1)
    ax.view_init(30, 30)
    ls = LightSource(-20, 45)
    ax.voxels(color_voxels, facecolors=color_voxels, edgecolor="k", lightsource=ls)
    plt.savefig(figname, bbox_inches="tight")
    plt.close()


if __name__ == "__main__":
    input_str = "\n".join(
        [
            "9 9 1",
            "9 16",
            "2 2 0",
            "2 4 0",
            "2 6 0",
            "4 2 0",
            "4 4 0",
            "4 6 0",
            "6 2 0",
            "6 4 0",
            "6 6 0",
            "0 0 0",
            "0 2 0",
            "0 4 0",
            "0 6 0",
            "0 8 0",
            "2 0 0",
            "2 8 0",
            "4 0 0",
            "4 8 0",
            "6 0 0",
            "6 8 0",
            "8 0 0",
            "8 2 0",
            "8 4 0",
            "8 6 0",
            "8 8 0",
            "8 3",
            "2",
            "0 1",
            "6",
            "2 2 2 0",
            "2 1 2 0",
            "1 1 2 0",
            "1 1 3 0",
            "1 2 3 0",
            "1 2 4 0",
            "2",
            "1 2",
            "6",
            "2 2 4 0",
            "2 1 4 0",
            "2 1 5 0",
            "2 2 5 0",
            "1 2 5 0",
            "1 2 6 0",
            "2",
            "2 3",
            "8",
            "2 2 6 0",
            "2 3 6 0",
            "1 3 6 0",
            "1 3 5 0",
            "1 3 4 0",
            "1 3 3 0",
            "1 4 3 0",
            "1 4 2 0",
            "2",
            "3 4",
            "6",
            "3 4 2 0",
            "3 3 2 0",
            "2 3 2 0",
            "2 3 3 0",
            "2 4 3 0",
            "2 4 4 0",
            "2",
            "4 5",
            "6",
            "3 4 4 0",
            "3 3 4 0",
            "2 3 4 0",
            "2 3 5 0",
            "2 4 5 0",
            "2 4 6 0",
            "2",
            "5 6",
            "8",
            "3 4 6 0",
            "3 5 6 0",
            "2 5 6 0",
            "2 5 5 0",
            "2 5 4 0",
            "2 5 3 0",
            "2 6 3 0",
            "2 6 2 0",
            "2",
            "6 7",
            "8",
            "3 6 2 0",
            "3 5 2 0",
            "3 5 3 0",
            "3 5 4 0",
            "3 5 5 0",
            "3 6 5 0",
            "2 6 5 0",
            "2 6 4 0",
            "2",
            "7 8",
            "8",
            "3 6 4 0",
            "3 7 4 0",
            "2 7 4 0",
            "2 7 5 0",
            "2 7 6 0",
            "2 7 7 0",
            "2 6 7 0",
            "2 6 6 0",
        ]
    )
    grid = parse_output(input_str)

    vis(grid, "vis.png")
```
